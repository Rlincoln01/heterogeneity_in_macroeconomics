Backward induction with cubic splines and Euler residual checks.

```python
import sys
from pathlib import Path
import numpy as np
import pandas as pd

ROOT = Path.cwd().resolve().parent
PY_DIR = ROOT / "python"
if str(PY_DIR) not in sys.path:
    sys.path.append(str(PY_DIR))

# Also add repository root so local hetmacro imports are robust.
HET_REPO = None
for p in [ROOT, *ROOT.parents]:
    if (p / "hetmacro").exists():
        HET_REPO = p
        break
if HET_REPO is None:
    raise RuntimeError("Could not locate repository root containing 'hetmacro'.")
if str(HET_REPO) not in sys.path:
    sys.path.append(str(HET_REPO))

from problem4_lifecycle import Params, solve_lifecycle
```

```python
out = solve_lifecycle(Params())
summary = pd.DataFrame([
    {
        "max_abs_euler_employed": float(np.max(np.abs(out["euler_e"]))),
        "max_abs_euler_unemployed": float(np.max(np.abs(out["euler_u"]))),
        "terminal_assets": float(out["sim_assets"][-1]),
        "mean_consumption": float(np.mean(out["sim_consumption"]))
    }
])
summary
```

```python
tex_dir = ROOT / "tex"
tex_dir.mkdir(exist_ok=True)
summary.to_csv(tex_dir / "problem4_summary.csv", index=False)

pd.DataFrame({
    "a": out["aa"],
    "ce_t1": out["ce_t1"],
    "cu_t1": out["cu_t1"],
    "ap_e_t1": out["ap_e_t1"],
    "ap_u_t1": out["ap_u_t1"],
    "euler_e": out["euler_e"],
    "euler_u": out["euler_u"],
}).to_csv(tex_dir / "problem4_policies_t1.csv", index=False)

pd.DataFrame({
    "t": np.arange(1, len(out["sim_assets"]) + 1),
    "assets": out["sim_assets"],
    "consumption": out["sim_consumption"],
    "income": out["sim_income"],
    "employment": out["sim_employment"],
}).to_csv(tex_dir / "problem4_simulation.csv", index=False)

print("Saved tables to", tex_dir)
```
