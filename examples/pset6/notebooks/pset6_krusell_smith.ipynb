# Problem Set 6: The Krusell-Smith Model

**Three approaches to aggregate dynamics in incomplete markets:**

1. **SSJ** (Sequence Space Jacobian) — linear impulse responses via partial-equilibrium Jacobians + GE closure
2. **Complete Markets RBC** — representative agent benchmark (linear + nonlinear projection)
3. **Krusell-Smith Algorithm** — original nonlinear simulation-based approach with law-of-motion iteration

All three methods use the same calibration and share the Aiyagari steady state.

## 1. Setup & Calibration

```python
import sys, os, time
import numpy as np
import matplotlib.pyplot as plt

# Add hetmacro to path
sys.path.insert(0, os.path.abspath("../../.."))

from hetmacro.markov import rouwenhorst, stationary_distribution as stat_dist
from hetmacro.grids import make_grid_1d
from hetmacro.steady_state import household_steady_state
from hetmacro.models.aiyagari import firm_k_demand, firm_wage
from hetmacro.optimize import brentq
from hetmacro.interpolation import get_lottery

plt.rcParams.update({
    "figure.figsize": (10, 6),
    "axes.grid": True,
    "font.size": 13,
    "lines.linewidth": 2.5,
})

print("Imports OK")
```

```python
# ── Calibration (matching Matlab pset6) ──────────────────────────────────
beta   = 0.99
theta  = 1.5       # CRRA
rho_e  = 0.99      # idiosyncratic persistence
sigma_e = 0.10     # idiosyncratic volatility
alpha  = 0.35      # capital share
delta  = 0.02      # depreciation
rho_z  = 0.95      # aggregate TFP persistence
sigma_z = 0.007    # aggregate TFP volatility

eis = 1.0 / theta  # elasticity of intertemporal substitution

# Grid sizes
nk = 151    # individual assets
ne = 11     # idiosyncratic productivity states
nK = 11     # aggregate capital (KS grid)
nz = 11     # aggregate TFP states

# ── Income process ───────────────────────────────────────────────────────
e_log, Qee = rouwenhorst(ne, rho_e, sigma_e)
egrid = np.exp(e_log)
pi_e = stat_dist(Qee)
egrid = egrid / np.dot(pi_e, egrid)  # normalise E[e] = 1

# ── TFP process ──────────────────────────────────────────────────────────
z_log, Qzz = rouwenhorst(nz, rho_z, sigma_z)
zgrid = z_log   # log deviations from Z = 1

# ── Asset grid (quadratic spacing, matching Matlab) ──────────────────────
kmin, kmax = 0.0, 2500.0
t_grid = np.linspace(0, 1, nk)
kgrid = kmin + (kmax - kmin) * t_grid**2

print(f"Calibration: beta={beta}, theta={theta}, alpha={alpha}, delta={delta}")
print(f"Income:  ne={ne}, rho_e={rho_e}, sigma_e={sigma_e}")
print(f"TFP:     nz={nz}, rho_z={rho_z}, sigma_z={sigma_z}")
print(f"Assets:  nk={nk}, kmin={kmin}, kmax={kmax}")
print(f"E[e] = {np.dot(pi_e, egrid):.6f}")
```

## 2. Aiyagari Steady State

Solve the standard Aiyagari GE: find $r^*$ such that household asset supply equals firm capital demand under Cobb-Douglas technology with $L=1$.

```python
# ── Solve Aiyagari GE steady state ───────────────────────────────────────

def excess(r):
    K_d = firm_k_demand(r, alpha, delta)
    w = firm_wage(K_d, alpha)
    ss_trial = household_steady_state(Qee, kgrid, w * egrid, r, beta, eis)
    return ss_trial["A"] - K_d

r_upper = 1.0 / beta - 1.0 - 1e-4
t0 = time.time()
r_star = brentq(excess, 0.001, r_upper, tol=1e-8)
K_d = firm_k_demand(r_star, alpha, delta)
w_star = firm_wage(K_d, alpha)
ss = household_steady_state(Qee, kgrid, w_star * egrid, r_star, beta, eis)
Kss = ss["A"]
elapsed = time.time() - t0

print(f"r*     = {r_star:.6f}   ({r_star*100:.3f}%)")
print(f"K_ss   = {Kss:.4f}")
print(f"W_ss   = {w_star:.4f}")
print(f"Y_ss   = {K_d**alpha:.4f}")
print(f"C_ss   = {ss['C']:.4f}")
rc = K_d**alpha - delta * K_d - ss["C"]
print(f"Resource constraint residual: {rc:.4e}")
print(f"Solved in {elapsed:.1f}s")
```

## 3. SSJ Impulse Responses

Compute impulse responses of aggregate capital to a TFP shock using the Sequence Space Jacobian method with Cobb-Douglas GE closure. Then simulate a capital path from the IRF coefficients.

```python
from hetmacro.models.krusell_smith import ssj_ge_irfs, simulate_from_ssj

T_irf = 300
t0 = time.time()
irfs = ssj_ge_irfs(ss, egrid, alpha, delta, rho_z, T=T_irf)
elapsed = time.time() - t0

print(f"SSJ IRFs computed in {elapsed:.1f}s (T={T_irf})")
print(f"Peak dlogK response: {irfs['dlogK'].max():.4f} at t={np.argmax(irfs['dlogK'])}")

# Plot IRF
fig, ax = plt.subplots(1, 1, figsize=(10, 5))
ax.plot(irfs["dlogK"][:200] * 100, label="dlog K")
ax.plot(irfs["dlogZ"][:200] * 100, label="dlog Z", ls="--", alpha=0.7)
ax.set_xlabel("Periods after shock")
ax.set_ylabel("% deviation from SS")
ax.set_title("SSJ Impulse Responses to Unit TFP Innovation")
ax.legend()
plt.tight_layout()
plt.show()
```

```python
# ── All GE IRFs from SSJ solution ────────────────────────────────────────
# Firm FOCs with K lagged one period (K_{t-1} enters production at t):
#   Y_t = Z_t * K_{t-1}^alpha   =>  dlogY_t = dlogZ_t + alpha*dlogK_{t-1}
#   R_t = alpha*Z_t*K_{t-1}^(alpha-1)  =>  dlogR_t = dlogZ_t + (alpha-1)*dlogK_{t-1}
#   W_t = (1-alpha)*Z_t*K_{t-1}^alpha  =>  dlogW_t = dlogZ_t + alpha*dlogK_{t-1}
# Consumption from resource constraint:
#   K_t = (1-delta)*K_{t-1} + X_t  =>  X_ss*dlogX_t = K_ss*(dlogK_t - (1-delta)*dlogK_{t-1})
#   C_ss*dlogC_t = Y_ss*dlogY_t - X_ss*dlogX_t

dlogZ = irfs['dlogZ']
dlogK = irfs['dlogK']

# Lagged capital (K_{t-1}); at t=0 the economy is at SS so dlogK_{-1} = 0
dlogK_lag = np.concatenate([[0.0], dlogK[:-1]])

# Steady-state levels
K_ss = ss['A']
r_ss = ss['r']
R_ss = r_ss + delta
W_ss = (1 - alpha) * K_ss**alpha      # at Z=1, N=1
Y_ss = K_ss**alpha
C_ss = Y_ss - delta * K_ss
X_ss = delta * K_ss

# IRFs from firm FOCs
dlogY = dlogZ + alpha * dlogK_lag
dlogR = dlogZ + (alpha - 1) * dlogK_lag   # log-dev of gross rental rate R = r+delta
dlogW = dlogZ + alpha * dlogK_lag

# IRFs from resource constraint
dlogX = K_ss * (dlogK - (1 - delta) * dlogK_lag) / X_ss
dlogC = (Y_ss * dlogY - X_ss * dlogX) / C_ss

# ── Plot: 3-panel layout ──────────────────────────────────────────────────
# Panel 1: Capital dynamics  (K, X)
# Panel 2: Output and consumption  (Y, C)
# Panel 3: Prices  (W, r)
T_plot = 100
tt = np.arange(T_plot)

fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# Panel 1 — Capital & investment
axes[0].plot(tt, dlogK[:T_plot] * 100, label='Capital ($K$)',     color='tab:blue',   lw=2)
axes[0].plot(tt, dlogX[:T_plot] * 100, label='Investment ($X$)',  color='tab:red',    lw=2, ls='--')
axes[0].axhline(0, color='k', lw=0.5, ls=':')
axes[0].set_title('Capital Dynamics', fontsize=11)
axes[0].set_xlabel('Periods after shock')
axes[0].set_ylabel('% dev from SS')
axes[0].legend(fontsize=9)

# Panel 2 — Output & consumption
axes[1].plot(tt, dlogY[:T_plot] * 100, label='Output ($Y$)',      color='tab:green',  lw=2)
axes[1].plot(tt, dlogC[:T_plot] * 100, label='Consumption ($C$)', color='tab:orange', lw=2, ls='--')
axes[1].axhline(0, color='k', lw=0.5, ls=':')
axes[1].set_title('Output & Consumption', fontsize=11)
axes[1].set_xlabel('Periods after shock')
axes[1].set_ylabel('% dev from SS')
axes[1].legend(fontsize=9)

# Panel 3 — Prices
axes[2].plot(tt, dlogW[:T_plot] * 100, label='Wage ($W$)',          color='tab:purple', lw=2)
axes[2].plot(tt, dlogR[:T_plot] * 100, label='Rental rate ($R$)',   color='tab:gray',   lw=2, ls='--')
axes[2].axhline(0, color='black', lw=1, ls='-')
axes[2].set_title('Prices', fontsize=11)
axes[2].set_xlabel('Periods after shock')
axes[2].set_ylabel('% dev from SS')
axes[2].legend(fontsize=9)

fig.suptitle('SSJ Impulse Responses to Unit TFP Innovation', fontsize=13)
plt.tight_layout()
plt.savefig('figures/IRFs_KS.png', dpi=150, bbox_inches='tight')
plt.show()

print("Peak responses (% dev from SS):")
for name, irf in [('K', dlogK), ('X', dlogX), ('Y', dlogY), ('C', dlogC), ('W', dlogW), ('R', dlogR)]:
    pk = irf[np.argmax(np.abs(irf))]
    t_pk = np.argmax(np.abs(irf))
    print(f"  {name}: {pk*100:+.3f}%  at t={t_pk}")
```

```python
# ── Simulate K path from SSJ IRF coefficients ───────────────────────────

np.random.seed(0)      # reproducible; matches Matlab rng('Default')
Tsim = 5100            # first 100 periods for burn-in
eps = np.random.randn(Tsim) * sigma_z

K_ssj, Z_sim = simulate_from_ssj(irfs["Kt_ssj"], eps, rho_z)

burn = 100
print(f"SSJ simulation: Tsim={Tsim}, burn-in={burn}")
print(f"  std(K)  = {np.std(K_ssj[burn:])*100:.2f}%")
print(f"  corr(K, Z) = {np.corrcoef(K_ssj[burn:], Z_sim[burn:])[0,1]:.4f}")
print(f"  K range: [{K_ssj.min():.4f}, {K_ssj.max():.4f}]")
```

## 5. Krusell-Smith Algorithm

The original KS algorithm iterates between:
1. **Inner loop**: solve the household problem on the 4D state $(k, K, e, Z)$ via EGM. For each aggregate state pair $(K_j, Z_m)$, the household faces a standard 1D savings problem, so the existing `backward_egm` machinery applies directly.
2. **Simulation**: forward-simulate the cross-sectional distribution to get a new $K$ path
3. **Projection**: re-estimate $K' = \Gamma(K, Z)$ from the simulated path using Chebyshev regression
4. **Damping**: update $K_t \leftarrow K_t + \text{step} \cdot (K_t^{\text{new}} - K_t)$

The initial guess for the $K$ path comes from the SSJ simulation above.

```python
from hetmacro.models.krusell_smith import solve_krusell_smith

# ── KS grids ────────────────────────────────────────────────────────────
zmin, zmax = zgrid[0], zgrid[-1]
Kmin, Kmax = 2 * zmin, 2 * zmax       # K grid range in log deviations
Kgrid_agg = np.linspace(Kmin, Kmax, nK)

params = {
    "beta": beta, "theta": theta,
    "alpha": alpha, "delta": delta,
    "Kss": Kss,
}

print(f"KS state space: {nk} x {nK} x {ne} x {nz} = {nk*nK*ne*nz:,} points")
print(f"K grid (log dev): [{Kmin:.4f}, {Kmax:.4f}]")
print(f"Z grid (log):     [{zmin:.4f}, {zmax:.4f}]")
```

```python
# ── Run KS outer loop ────────────────────────────────────────────────────
# EGM-based: each outer iteration takes ~6s (vs minutes with VFI).
# Typically converges in 10-20 outer iterations.
# Set RUN_KS = False to skip.

RUN_KS = True
MAX_OUTER = 20

if RUN_KS:
    t0 = time.time()
    ks_result = solve_krusell_smith(
        ss, params, Z_sim, K_ssj, Qee, Qzz,
        kgrid, Kgrid_agg, egrid, zgrid,
        ncheb_K=3, ncheb_Z=3,
        step=0.5,
        tol_outer=1e-6,
        max_outer=MAX_OUTER,
        burn_in=burn,
        verbose=True,
        tol=1e-7,
        max_iter=500,
    )
    elapsed = time.time() - t0

    K_ks = ks_result.K_path
    print(f"\nTotal KS time: {elapsed:.0f}s ({elapsed/60:.1f} min)")
    print(f"Converged: {ks_result.converged} ({ks_result.n_iterations} outer iters)")
    print(f"Final R² = {ks_result.Rsq:.6f}")
else:
    K_ks = K_ssj  # fallback
    print("Skipped KS algorithm (using SSJ path as proxy)")
```

## Discussion: Approximate Aggregation

The key finding of Krusell and Smith (1998) is **approximate aggregation**: the law of motion $K' = \Gamma(K, Z)$ achieves $R^2 > 0.999$, meaning that aggregate capital and TFP are nearly sufficient statistics for forecasting future aggregate capital. The full cross-sectional distribution adds almost no additional predictive power.

This implies that, for aggregate dynamics, the incomplete markets economy with heterogeneous agents behaves almost identically to a representative agent economy. The cross-sectional distribution shifts in response to aggregate shocks, but these shifts have negligible effects on aggregate capital accumulation.

The SSJ method exploits this by linearising around the steady state distribution, capturing the same near-aggregation result at a fraction of the computational cost.

## 7. Bayesian Estimation of the TFP Process

Estimate $(\rho_z, \sigma_z)$ from HP-filtered US GDP using the SSJ model as the structural mapping.

**Observation equation:** The model implies output dynamics $y_t = \sum_{s=0}^{T-1} \Gamma_s \, \varepsilon_{t-s}$ where $\Gamma_s$ is the output IRF at horizon $s$ and $\varepsilon_t \sim N(0, \sigma_z^2)$. The data is $y_t^{\text{data}} = y_t^{\text{model}} + \eta_t$ with $\eta_t \sim N(0, \sigma_{\text{meas}}^2)$.

**Likelihood:** Gaussian, computed from model-implied autocovariances via Toeplitz covariance + Cholesky factorisation.

**Key efficiency trick:** The steady state and GE Jacobians do NOT depend on $(\rho_z, \sigma_z)$. We precompute the output IRF matrix $G_{YZ}$ once, then each MCMC iteration is just one matrix-vector multiply + autocovariance + likelihood (~0.3ms per draw).

```python
# ── Download US Real GDP from FRED ───────────────────────────────────────
# GDPC1: Real GDP (quarterly, chained 2017 dollars). Sample ends 2019Q4 (pre-pandemic).
# HP filter (lambda=1600) extracts the cyclical component.

import pandas as pd

try:
    from pandas_datareader import data as pdr
    gdp_raw = pdr.DataReader("GDPC1", "fred", start="1950-01-01", end="2019-12-31")
    gdp = gdp_raw["GDPC1"].dropna()
    log_gdp = np.log(gdp)

    try:
        from statsmodels.tsa.filters.hp_filter import hpfilter
        cycle, trend = hpfilter(log_gdp, lamb=1600)
        y_data = cycle.values
        print(f"Downloaded GDPC1: {len(y_data)} quarters, HP-filtered")
    except ImportError:
        t_idx = np.arange(len(log_gdp))
        coef = np.polyfit(t_idx, log_gdp.values, 1)
        y_data = log_gdp.values - np.polyval(coef, t_idx)
        print(f"Downloaded GDPC1: {len(y_data)} quarters, linearly detrended")

    y_dates = gdp.index
    freq = "quarterly"
    print(f"  {y_dates[0].date()} to {y_dates[-1].date()} (pre-pandemic)")
except Exception as e:
    print(f"FRED download failed ({e}), generating synthetic data")
    np.random.seed(42)
    T_data = 280
    rho_true, sigma_true = 0.95, 0.007
    y_data = np.zeros(T_data)
    for t in range(1, T_data):
        y_data[t] = rho_true * y_data[t - 1] + np.random.randn() * sigma_true
    y_dates = pd.date_range("1950Q1", periods=T_data, freq="QS")
    freq = "quarterly (synthetic)"
    print(f"  {T_data} quarters, rho_true={rho_true}, sigma_true={sigma_true}")

print(f"\nObservable: {len(y_data)} obs ({freq})")
print(f"  std = {np.std(y_data)*100:.3f}%,  autocorr = {np.corrcoef(y_data[:-1], y_data[1:])[0,1]:.4f}")

fig, ax = plt.subplots(figsize=(10, 4))
ax.plot(y_dates, y_data * 100)
ax.axhline(0, color="k", lw=0.5)
ax.set_ylabel("log GDP cycle (%)")
ax.set_title("US Real GDP — Cyclical Component (HP filter, $\\lambda=1600$, pre-pandemic)")
plt.tight_layout()
plt.show()
```

```python
# ── SSJ estimation infrastructure ────────────────────────────────────────
from scipy.linalg import toeplitz, cholesky, solve_triangular
from scipy.stats import beta as beta_dist, invgamma
from tqdm.auto import trange

from hetmacro.ssj import build_ge_matrices


def build_output_irf_matrix(ss, irfs, alpha_val, delta_val, T):
    """Precompute the output IRF matrix G_YZ (parameter-independent).

    G_YZ maps a TFP shock path dlogZ -> output IRF dlogY:
        dlogY = G_YZ @ dlogZ

    Since dlogY = dlogZ + alpha * dlogK and dlogK = solve(H_U, -H_Z @ dlogZ),
    we have G_YZ = I + alpha * inv(H_U) @ (-H_Z) = I + alpha * G_KZ.

    This matrix does NOT depend on (rho_z, sigma_z).
    """
    r_ss = ss["r"]
    K_ss = ss["A"]
    R_ss = r_ss + delta_val
    W_ss = (1.0 - alpha_val) * K_ss ** alpha_val

    J_r = irfs["J_r"]
    J_W = irfs["J_W"]

    # Lag operator: prices at t depend on K chosen at t-1
    L = np.zeros((T, T))
    L[1:, :-1] = np.eye(T - 1)

    price_elast_K = {
        "r_shock": R_ss * (alpha_val - 1) * L,
        "W_shock": W_ss * alpha_val * L,
    }
    price_elast_Z = {
        "r_shock": R_ss * np.eye(T),
        "W_shock": W_ss * np.eye(T),
    }

    H_U, H_Z = build_ge_matrices(
        Js_A={"r_shock": J_r, "W_shock": J_W},
        price_elasticities_K=price_elast_K,
        price_elasticities_Z=price_elast_Z,
        K_ss=K_ss,
        T=T,
    )

    # G_KZ = inv(H_U) @ (-H_Z)  →  dlogK = G_KZ @ dlogZ
    G_KZ = np.linalg.solve(H_U, -H_Z)
    # dlogY = dlogZ + alpha * dlogK = (I + alpha * G_KZ) @ dlogZ
    G_YZ = np.eye(T) + alpha_val * G_KZ
    return G_YZ


def model_autocovariances(G_YZ, rho_z, sigma_z, sigma_meas, T, n_lags):
    """Model-implied output autocovariances for given (rho_z, sigma_z).

    Returns gamma[h] for h = 0, ..., n_lags-1.
    """
    dlogZ = rho_z ** np.arange(T)
    dlogY_irf = G_YZ @ dlogZ                         # output IRF to unit innovation
    full_corr = np.correlate(dlogY_irf, dlogY_irf, mode="full")
    gamma = sigma_z ** 2 * full_corr[T - 1 : T - 1 + n_lags]
    gamma[0] += sigma_meas ** 2
    return gamma


def gaussian_log_likelihood(y, gamma):
    """Gaussian log-likelihood from Toeplitz autocovariance structure."""
    T_obs = len(y)
    V = toeplitz(gamma[:T_obs])
    try:
        L = cholesky(V, lower=True)
    except np.linalg.LinAlgError:
        return -np.inf
    z = solve_triangular(L, y, lower=True)
    log_det = 2.0 * np.sum(np.log(np.diag(L)))
    return -0.5 * (log_det + np.dot(z, z) + T_obs * np.log(2 * np.pi))


def log_prior_est(rho, sigma):
    """Log-prior: rho ~ Beta(20,2), sigma ~ InvGamma(3, 0.02)."""
    if not (0 < rho < 1) or sigma <= 0:
        return -np.inf
    return beta_dist.logpdf(rho, 20, 2) + invgamma.logpdf(sigma, a=3, scale=0.02)


def log_posterior_est(rho, sigma, y, G_YZ, T, sigma_meas):
    """Unnormalised log-posterior = log-likelihood + log-prior."""
    lp = log_prior_est(rho, sigma)
    if not np.isfinite(lp):
        return -np.inf
    gamma = model_autocovariances(G_YZ, rho, sigma, sigma_meas, T, len(y))
    ll = gaussian_log_likelihood(y, gamma)
    if not np.isfinite(ll):
        return -np.inf
    return ll + lp


def run_mcmc(y, G_YZ, T, sigma_meas=1e-4,
             n_draws=50_000, burn_in_frac=0.5,
             init=(0.9, 0.01), proposal_scale=(0.005, 0.0005), seed=123):
    """Random-walk Metropolis-Hastings for (rho_z, sigma_z)."""
    rng = np.random.default_rng(seed)
    n_burn = int(n_draws * burn_in_frac)

    chain = np.empty((n_draws, 2))
    chain[0] = init
    lp_curr = log_posterior_est(*init, y, G_YZ, T, sigma_meas)

    n_accept = 0
    for i in trange(1, n_draws, desc="MCMC"):
        rho_prop = chain[i - 1, 0] + rng.normal() * proposal_scale[0]
        sig_prop = chain[i - 1, 1] + rng.normal() * proposal_scale[1]

        lp_prop = log_posterior_est(rho_prop, sig_prop, y, G_YZ, T, sigma_meas)

        if np.log(rng.uniform()) < lp_prop - lp_curr:
            chain[i] = [rho_prop, sig_prop]
            lp_curr = lp_prop
            n_accept += 1
        else:
            chain[i] = chain[i - 1]

    posterior = chain[n_burn:]
    return {
        "chain": chain,
        "posterior": posterior,
        "accept_rate": n_accept / (n_draws - 1),
        "rho_mean": posterior[:, 0].mean(),
        "rho_std": posterior[:, 0].std(),
        "sigma_mean": posterior[:, 1].mean(),
        "sigma_std": posterior[:, 1].std(),
    }

print("Estimation infrastructure defined")
```
